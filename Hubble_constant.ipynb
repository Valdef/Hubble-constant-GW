```python
from uncertainties import unumpy

H0 = np.array([event.H0 for event in event.values() if not np.isnan(event.H0)]) # get all H0
H0_low = np.array([event.low_uncertainty for event in event.values() if not np.isnan(event.low_uncertainty)]) # get all low uncertainty
H0_high = np.array([event.high_uncertainty for event in event.values() if not np.isnan(event.high_uncertainty)]) # get all high uncertainty

H0_w_low_err = unumpy.uarray(H0,abs(H0_low))
H0_w_high_err = unumpy.uarray(H0,H0_high)

# get mean of value 
mean_low_err = np.mean(H0_w_low_err)
mean_high_err = np.mean(H0_w_high_err)

# get specific value and uncertainty
mean_H0 = unumpy.nominal_values(mean_low_err)
mean_H0_low = unumpy.std_devs(mean_low_err)
mean_H0_high = unumpy.std_devs(mean_high_err)
    
print(f"H0: {mean_H0:.3f} (+{mean_H0_high:.3f} -{mean_H0_low:.3f}) Mpc")
```

# Plot graph of cumulative mean

```python
import matplotlib.pyplot as plt
from uncertainties import ufloat

# Calculate cumulative means and cumulative errors of H0
H0_mean_low = [np.mean(H0_w_low_err[:i + 1]) for i in range(len(H0_w_low_err))]
H0_mean_high = [np.mean(H0_w_high_err[:i + 1]) for i in range(len(H0_w_high_err))]

cumul_means = np.array([x.n for x in H0_mean_low])
cumul_err_low = np.array([x.s for x in H0_mean_low])
cumul_err_high = np.array([x.s for x in H0_mean_high])

x_positions = np.arange(1, len(event) + 1) # get points between 0 and 23 to plot x-axis


plt.figure(figsize=(10, 6))

# Plot cumulative means with error bars
if galaxy_survey == 'GLADE+ and SDSS(Mean)':
    plt.errorbar(x_positions[:18], cumul_means[:18], yerr=[cumul_err_low[:18],cumul_err_high[:18]], 
                 fmt='o', color='black', capsize=5, label='Cumulative Mean GLADE+')
    
    # Fill the area between error points
    plt.fill_between(x_positions[:18], cumul_means[:18] - cumul_err_low[:18], cumul_means[:18] + cumul_err_high[:18], 
                     color='dimgray', alpha=0.2)
    
    plt.errorbar(x_positions[18:], cumul_means[18:], yerr=[cumul_err_low[18:],cumul_err_high[18:]], 
                 fmt='o', color='midnightblue', capsize=5, label='Cumulative Mean SDSS (Mean)')
    # Fill the area between error points
    plt.fill_between(x_positions[17:], cumul_means[17:] - cumul_err_low[17:], cumul_means[17:] + cumul_err_high[17:], 
                     color='midnightblue', alpha=0.2)
    
else:
    plt.errorbar(x_positions, cumul_means, yerr=[cumul_err_low,cumul_err_high], 
                 fmt='o', color='black', capsize=5, label=f'Cumulative Mean {galaxy_survey}')
    # Fill the area between error points
    plt.fill_between(x_positions, cumul_means - cumul_err_low, cumul_means + cumul_err_high, 
                     color='dimgray', alpha=0.2)


# Uncomment to see the different distances associated with event
#plt.vlines(15,0,140, color='slategrey', linestyle='--', label='1042Mpc')
#plt.vlines(19,0,140, color='black', linestyle='--', label='2500Mpc')

# plot line for H0 Cepheid measurement
plt.hlines(73.3, 1, 23, label='$H_{0, Ceph}$') # Cepheid
plt.fill_between(x_positions, 73.3-1, 73.3 + 1, color='blue', alpha=0.2)

# plot line for H0 CMB measurement
plt.hlines(67.4, 1, 23, color='green', label='$H_{0, CMB}$') # CMB
plt.fill_between(x_positions, 67.4 - 0.5, 67.4 + 0.5, color='green', alpha=0.2)


plt.xticks(x_positions, ha='right')
plt.xlabel('Number of Event')
plt.ylabel('Hubble Constant $H_0$ $(kms^{-1}Mpc^{-1})$')
plt.title('Cumulative Mean of $H_0$ Measurements from Gravitational Wave Events')
plt.legend()
plt.grid(True, alpha=0.5)
plt.tight_layout()
plt.show()
```
